```python
import glob, pandas as pd, numpy as np, matplotlib.pyplot as plt, time

nsga = sorted(glob.glob("./outputs/front_nsga2_*.csv"))
dfa  = sorted(glob.glob("./outputs/front_dfa_*.csv"))
if not nsga: raise FileNotFoundError("No se encontró front_nsga2_*.csv en ./outputs")
if not dfa:  raise FileNotFoundError("No se encontró front_dfa_*.csv en ./outputs")

f_nsga = nsga[-1]; f_dfa = dfa[-1]
dnsga = pd.read_csv(f_nsga); ddfa = pd.read_csv(f_dfa)
F_nsga = dnsga[["Energia_kW","Xo_prom_ultimos"]].to_numpy()
F_dfa  = ddfa[["Energia_kW","Xo_prom_ultimos"]].to_numpy()

plt.figure(figsize=(7.5,4.5))
plt.scatter(F_nsga[:,0], F_nsga[:,1], s=36, facecolor="#1f77b4", edgecolor="black", alpha=0.95, label="Frente - NSGAII")
plt.scatter(F_dfa[:,0],  F_dfa[:,1],  s=28, facecolor="#ff7f0e", edgecolor="black", alpha=0.9,  label="Frente - DFA")
plt.xlabel("Gasto energético (kW)")
plt.ylabel("Contenido de agua (kg/kg)")
plt.title("Comparación de frentes de Pareto")
plt.legend(); plt.tight_layout()
stamp = time.strftime("%Y%m%d-%H%M%S")
plt.savefig(f'./outputs/pareto_comparacion_{stamp}.png', dpi=200); plt.savefig(f'./outputs/pareto_comparacion_{stamp}.svg')
plt.show()
```
